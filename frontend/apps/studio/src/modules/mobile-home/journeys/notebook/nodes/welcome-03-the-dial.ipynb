# The Dial

Numbers in Python work like a dial — change the value and the output changes.

Run the cell below to see the current settings.

```python
volume = 5
print("Volume is set to:", volume)
print("Twice as loud would be:", volume * 2)
```

## Turn the dial

Change `volume` to any number between 1 and 10 and run the cell again. Notice how the output changes.

```python
speed = 3
distance = 100
time = distance / speed
print("At speed", speed, "it takes", time, "seconds to travel", distance, "meters.")
```
